## Download a subset from MOFDB and Quick Cluster

In this tutorial, we see how to download a subset of structures from the MOFDB database using a wrapper function in mofography. MOFDB is a large database of nanoporous materials, including MOFs and zeolites that is hosted by Northwestern University. 
https://mof.tech.northwestern.edu/

For a lot of the materials workflows, it is very convenient to have a subset of structures to be used as a starting point, and to be saved as a pandas dataframe, that some of the textural properties of the material along with its crystallographic information. However, downloading large datasets can be cumbersome. Hence, mofography provides a wrapper that can do this asynchronously, meaning, the pages and pages of your search results will be downloaded in parallel, saving a lot of time. Note that at the time of writing, MOFDB itself has a python API that does this, https://github.com/n8ta/mofdb-client, the comparison of speeds of the two codes is not tested, but you are welcome to test it out as well. 

*Note that in order to avoid excessive strain on the MOFDB website, it is recommended you save the downloaded dataset locally (we will show this at the end) instead of downloading it every time.*

Once we have the dataset, we can create ASE atoms objects from it, pick one or more at random and perform a mofography workflo. In this example, we simply perform the quick cluster workflow, but feel free to build one of your own. If you don't remember what the Quick Cluster workflow does, please go check out the Quick Cluster notebook in the tutorials.

## Set paths and import mofography


```python
import mofography as mgr
```

Now grabbing data from MOFDB using mofography is very easy. There are many options by which you can filter the data, you can find them here: https://mof.tech.northwestern.edu/api

As an example, let's download all the materials with void fraction between 0.7 and 0.8, and with a largest cavity diameter between 12 and 13 angstroms from the CoREMOF-2019 database. Granted this is only small subset, selected purely for demonstration purposes.
It takes about 4 mins on a regular day with this set of parameters

```python
# get_data_from_mofdb is async: it must be awaited (top-level await in a notebook,
# asyncio.run(...) in a script)
df = await mgr.get_data_from_mofdb(params={'vf_min': 0.7, 'vf_max': 0.75, 'lcd_min': 12, 'lcd_max': 14, 'database': 'CoREMOF 2019'})
```

## Create ASE atoms objects
Since all the calculations in mofography start with an ASE atoms, object, let's convert make a list of ASE atoms objects for all the materiasl in this dataframe. 

```python
atoms_list = mgr.atoms_from_mofdb_df(df)
```

We can now do a workflow on one or more of this atoms objects, parallelize them whichever way we want and whatnot. Let's save that for another tutorial. For the time being, let's just pick one at random and run the quick cluster workflow on it and visualize

If you don't remember what the quick cluster workflow does, pleae go back and refer to the Quick Cluster notebook in the docs.

```python
import random 
index = random.randint(0,len(atoms_list)-1)
mat_atoms = atoms_list[index]
mat_updated = mgr.update_with_covalent_bonds(mat_atoms)

```

## Segment the pore space

Now run the standard workflow from the [Quick start](quick_start.qmd) on this atoms object: distance grid, periodic segmentation, then plot.

```python
import numpy as np
spacing = 0.5
dgrid = mgr.dgrid_from_atoms_rust(mat_atoms, spacing=spacing).astype(np.float32)
seg = mgr.do_periodic_segmentation_rust(
    dgrid, mat_atoms, spacing=spacing,
    mask_thickness=0.05, h=0.2, min_distance=1.5, threshold_abs=1.5,
    compute_connections=True, return_dgrid=True, disable_tqdm=True,
)
print(f"Pockets (PBC-merged): {len(seg['maxima_pbc'])}")
```

Let's visualize the result. Change the 'renderer' to 'notebook_connected' for an interactive plot in the notebook, or to 'browser' for a browser tab.

```python
fig = mgr.show_mof_plotly(seg, mat_atoms, stylize=True)
fig.show(renderer='png')
```
